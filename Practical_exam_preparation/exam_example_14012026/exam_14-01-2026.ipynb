## Data set and socio-technical context
In this exam you will work with **bicycle accident data** from Rotterdam. These data can be used to analyze the safety of bicycle infrastructure and identify high-risk areas. The data set contains information on 
bicycle accidents, including the severity of the accident (minor or severe), the location of the accident, and environmental factors such as weather conditions and road surface type. <br><br>

Columns in the data set include:
- `street_id`: Unique id for the street segment
- `oneway`: Whether the street segment is one-way (1) or two-way (0)
- `length`: Length of the street segment in meters
- `speed`: Avg cycling speed on the street segment in km/h
- `n_observations`: Number of data points for the street segment
- `n_severe`: Number of severe accidents on the street segment
- `n_minor`: Number of minor accidents on the street segment
- `rain_prop`: Weather condition during accident. Raining (1) or Dry (0), averaged across accidents
- `dark_prop`:  Light condition during accident. Dark (1) or Lit (0), averaged across accidents
- `street_type`: Type of street (categorical)
